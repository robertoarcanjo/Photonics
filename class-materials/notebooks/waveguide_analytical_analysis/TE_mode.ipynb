# TE Mode



```python
import numpy as np
import matplotlib.pyplot as plt
```

Equation for $b(V)$

$2V\sqrt(1-b) = m\pi + \arctan \left (\sqrt \frac{b + \delta}{1-b} \right) + \arctan \left (\sqrt \frac{b}{1-b} \right)$

```python
b = np.linspace(0,1,100)
m = [0,1,2,3]
delta = [0,2,4,500]
V = lambda m,b,delta: 0.5*(1/np.sqrt(1-b))*(m*np.pi+ np.arctan(np.sqrt((b+delta)/(1-b))) + np.arctan(np.sqrt(b/(1-b))))
V_data = np.zeros((len(m),len(b),len(delta)))
for i in range(len(m)):
    for j in range(len(b)):
        for k in range(len(delta)):
            V_data[i,j,k] = V(m[i],b[j],delta[k])


for i in range(len(m)):
    plt.plot(V_data[i,:,0],b,label=f'm={m[i]} delta=0')
    plt.plot(V_data[i,:,1],b,'--',label=f'm={m[i]} delta=2')
    plt.plot(V_data[i,:,2],b,':',label=f'm={m[i]} delta=4')
    plt.plot(V_data[i,:,3],b,'-.',label=f'm={m[i]} delta=INF')

plt.xlabel('V')
plt.ylabel('b')
plt.xlim(0,6)
plt.grid()
plt.show()  
```
