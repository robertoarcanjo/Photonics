# Operadores

```python
import scipy.sparse as sp
import numpy as np

def Dxx(N, dx):
    # Método da diferença central
    diagonals = [np.ones(N), -2 * np.ones(N), np.ones(N)]
    offsets = [-1, 0, 1]
    D2 = sp.diags(diagonals, offsets, shape=(N, N), format='csr')
    return D2 / (dx ** 2)

def Dyy(N, dy):
    # Método da diferença central
    diagonals = [np.ones(N), -2 * np.ones(N), np.ones(N)]
    offsets = [-4, 0, 4]
    D2 = sp.diags(diagonals, offsets, shape=(N, N), format='csr')
    return D2 / (dy ** 2)
```
